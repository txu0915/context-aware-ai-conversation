## How the tutorial examples are constructed

1. Start from a mathematically valid problem and an observable student artifact.
2. Select an error class and evidence policy before generating dialogue.
3. Use role-specialized agents to author three matched short, medium, and long dialogue scenarios.
4. Within each pair, keep the problem, misconception, and student profile fixed; only the context-aware tutor receives structured error evidence.
5. Score only visible teacher behavior and student evidence.
6. Use the resulting six sessions to connect the three sections into one traceable system story.

These agent-authored, tutorial-curated examples demonstrate the workflow and data contracts; they are not causal performance claims. The optional live cells let participants try the same workflow with current API models.
