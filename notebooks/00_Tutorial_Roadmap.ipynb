## How the tutorial examples are constructed

1. Start from a mathematically valid problem and an observable student artifact.
2. Select an error class and evidence policy before generating dialogue.
3. Create a matched pair: both tutors receive the same problem and student state, while only one receives structured error context.
4. Keep the student misconception and response style consistent across the pair.
5. Score only visible teacher behavior and student evidence.
6. Use the resulting examples to connect the five sections into one traceable system story.

The examples demonstrate the workflow and data contracts. The optional live cells let participants reproduce the same workflow with current API models.
